# SpotEdit on Qwen-Image-Edit-2511

Qwen-Image-Edit-2511 uses the **same** `QwenImageEditPlusPipeline` as 2509, so it reuses the `Qwen_image_edit_plus` SpotEdit module unchanged — only the repo id and the recommended sampling defaults differ (2511: `num_inference_steps=40`, `guidance_scale=1.0`, `true_cfg_scale=4.0`).

```python
%load_ext autoreload
%autoreload 2
```

```python
import torch
from diffusers import QwenImageEditPlusPipeline
from diffusers.utils import load_image
from Qwen_image_edit_plus import generate, SpotEditConfig
```

```python
pipe = QwenImageEditPlusPipeline.from_pretrained('Qwen/Qwen-Image-Edit-2511', torch_dtype=torch.bfloat16).to('cuda')
```

```python
image = load_image('./asset/dog.jpg')
image = image.resize((1024, 1024))
```

```python
config = SpotEditConfig(threshold=0.15, reset_steps=[11, 19, 27])
```

```python
res = generate(
    pipe,
    image = image,                 # single image, or a list for multi-reference editing
    prompt = 'add a scarf to the dog',
    config = config,
    num_inference_steps=40,
    guidance_scale=1.0,
)
res.images[0]
```
